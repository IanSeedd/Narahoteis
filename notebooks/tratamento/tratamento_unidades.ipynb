```python
import pandas as pd
import numpy as np
```

```python
df_antigo = pd.read_csv('../../dados/raw/unidades.csv', sep=';')
df_novo = pd.read_csv('../../dados/raw/unidades_novos.csv', sep=';')
```

```python
df = pd.concat(
    [df_antigo, df_novo],
    ignore_index=True
)
df
```

```python
df.info()
```

### Tratamentos de texto:

#### Tratamento de região:

```python
df['regiao'] = df['regiao'].str.strip().str.title()
```

```python
map_regiao = {
    'Cap': 'Capital',
}
df['regiao'] = df['regiao'].map(map_regiao).fillna(df['regiao'])
```

```python
df['regiao'].value_counts().reset_index()
```

```python
df['nome_unidade'].value_counts().reset_index()
```

#### Tratando estrelas:

```python
df['categoria_hotel'] = df['categoria_hotel'].str.strip().str.title()
```

```python
df['categoria_hotel'].value_counts().reset_index()
```

```python
map_cathotel = {
    'Cinco Estrelas': '5 Estrelas',
    'Três Estrelas': '3 Estrelas',
    '3*': '3 Estrelas',
}
df['categoria_hotel'] = df['categoria_hotel'].map(map_cathotel).fillna(df['categoria_hotel'])
```

```python
df['categoria_hotel'].value_counts().reset_index()
```

```python
df['num_quartos_total'] = df['num_quartos_total'].astype('Int64')
```

```python
df.info()
```
