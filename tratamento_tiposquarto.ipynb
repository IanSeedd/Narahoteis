```python
import pandas as pd
import numpy as np
```

```python
df_antigo = pd.read_csv('narahoteisbases/tipos_quarto.csv', sep=';')
df_novo = pd.read_csv('narahoteisbases_nova/tipos_quarto_novos.csv', sep=';')
```

```python
df = pd.concat(
    [df_antigo, df_novo],
    ignore_index=True
)
df
```

```python
df.info()
```

```python
df['valor_diaria_base'] = df['valor_diaria_base'].str.replace('.', '', regex=False).str.replace('R$', '', regex=False).str.replace(' ', '', regex=False).str.replace(',', '.', regex=False).astype(float)
```

```python
df['descricao'] = df['descricao'].str.strip().str.title()
```

```python
df = df.drop_duplicates(keep='first')
```
